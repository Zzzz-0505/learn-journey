```python
#写入
season = {
    "name": "spring",
    "months": ["March", "April", "May"],
    "description": "万物复苏的季节",
    "temperature": {
        "min": 10,
        "max": 25,
        "unit": "Celsius"
    },
    "activities": ["踏青", "赏花", "春游"],
    "isActive": True,
    "averageRainfall": 120.5,
    "note": None
}
with open("season.json", "w", encoding="utf-8") as f:
    json.dump(season, f,ensure_ascii=False,indent=2)
```

```python
#读取
with open('season.json','r') as f:
    loaded = json.load(f)
    type_isActive = loaded['isActive']
    type_note = loaded['note']
print(loaded)
print(type(loaded))
print(type(type_isActive))
print(type(type_note))
```

```python
#考虑中文
season = {
    "name": "spring",
    "months": ["March", "April", "May"],
    "description": "万物复苏的季节",
    "temperature": {
        "min": 10,
        "max": 25,
        "unit": "Celsius"
    },
    "activities": ["踏青", "赏花", "春游"],
    "isActive": True,
    "averageRainfall": 120.5,
    "note": None
}
with open("season2.json", "w", encoding="utf-8") as f:
    json.dump(season, f,ensure_ascii=True,indent=2)
```

```python
#dump()和dumps()的区别

data = {"name": "spring", "isActive": True}

with open("season.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)
    
```

```python
data = {"name": "spring", "isActive": True}

s = json.dumps(data, ensure_ascii=False, indent=2)
print(s)
print(type(s))
```
