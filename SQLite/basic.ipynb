```python
import sqlite3
```

```python
#连接数据库，没有则新建
conn = sqlite3.connect('test.db')

print ("数据库打开成功")
```

```python
#创建表
c = conn.cursor()
c.execute('''CREATE TABLE COMPANY
(ID INT PRIMARY KEY     NOT NULL,
NAME        TEXT    NOT NULL,
AGE         INT     NOT NULL,
ADDRESS     CHAR(50),
SALARY      REAL);
''')
print("数据表创建成功")
conn.commit()
conn.close()
```

```python
#INSERT 操作
conn = sqlite3.connect('test.db')
c = conn.cursor()
print("OPEN!")

c.execute("INSERT INTO COMPANY(ID,NAME,AGE,ADDRESS,SALARY) \
          VALUES(1,'a',32,'beijing',20000.00)")
c.execute("INSERT INTO COMPANY(ID,NAME,AGE,ADDRESS,SALARY) \
          VALUES(2,'b',25,'shanghai',30000.00)")
c.execute("INSERT INTO COMPANY(ID,NAME,AGE,ADDRESS,SALARY) \
          VALUES(3,'c',32,'guangzhou',40000.00)")
c.execute("INSERT INTO COMPANY(ID,NAME,AGE,ADDRESS,SALARY) \
          VALUES(4,'d',32,'hangzhou',50000.00)")
conn.commit()
print("数据插入成功!")
conn.close()
```

```python
#SELECT 操作
conn = sqlite3.connect('test.db')
c = conn.cursor()
print("open successful!")

cursor = c.execute("SELECT id,name,address,salary from COMPANY")
for row in cursor:
    print("ID= ",row[0])
    print("Name=",row[1])
    print("Address=",row[2])
    print("Salary=",row[3],"\n")
print("data operation successful!")
conn.close()
```

```python
#UPDATE 操作
conn = sqlite3.connect("test.db")
c = conn.cursor()
print("database open successful!")
c.execute("UPDATE COMPANY set SALARY = 25000.00 where ID =1")
conn.commit()
print("Total number of rows updated:",conn.total_changes)

cursor = conn.execute("SELECT id,name,address,salary from COMPANY")
for row in cursor:
    print("ID= ",row[0])
    print("Name=",row[1])
    print("Address=",row[2])
    print("Salary=",row[3],"\n")
print("data operation successful!")
conn.close()
```

```python
#DELETE 操作
conn = sqlite3.connect('test.db')
c = conn.cursor()
print("database opem successful!")

c.execute("DELETE from COMPANY where ID=2;")
conn.commit()
print("Total number of rows deleted :",conn.total_changes)

cursor =  conn.execute("SELECT id,name,address,salary from COMPANY")
for row in cursor:
    print(f"ID={row[0]}")
    print(f"NAME={row[1]}")
    print(f"ADDRESS={row[2]}")
    print(f"SALARY={row[3]}")
print("data operation successful!")
conn.close()
```
